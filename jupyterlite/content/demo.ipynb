## The pipe operator `|>`

`x |> f` is just `f(x)`, but written in natural left-to-right reading order:

```python
tup = (3, 4, 1, 5, 6)
tup |> sorted |> tuple
```

## Placeholders with `$`

`$` stands in for an argument and induces a function. Use it to pass extra keyword args, or anywhere in an expression:

```python
lst = [3, 4, 1, 5, 6]
lst |> sorted($, reverse=True)
```

```python
# find the position of 3 after sorting
lst |> sorted |> $.index(3)
```

## Pipelines as functions

If a placeholder appears in the *first* step, the whole pipeline becomes a function. Named placeholders (`$a`, `$b`) let you introduce more than one argument:

```python
hypotenuse = $a * $a + $b * $b |> $ ** 0.5
hypotenuse(3, 4)
```

## Macros

Built-in functional macros like `reduce`, `map`, and `filter` take their function in brackets — and inside those brackets you can drop the `f` from a quick lambda, so `reduce[$ * $]` folds with `lambda a, b: a * b`:

```python
# factorial of 4
range(1, 5) |> reduce[$ * $]
```

```python
# build a number from its decimal digits
[2, 3, 4] |> reduce[10*$ + $]
```

```python
# the quick-lambda macro `f` on its own
f[$a*$b + $b*$c + $a*$c](2, 3, 4)
```

## Brace blocks

A `macro{ ... }` body can be a multi-line statement block — loops, conditionals, intermediate assignments. The block's value is its final expression:

```python
5 |> f{
    acc = 0
    for i in range($):
        acc += i
    acc
}
```

## Forking and function composition

`fork` applies the same input to several pipelines and returns a tuple; `.>` composes functions left-to-right:

```python
5 |> fork{ $ + 1, $ * 2 }
```

```python
reverse = reversed .> list
[1, 2, 3] |> reverse
```
